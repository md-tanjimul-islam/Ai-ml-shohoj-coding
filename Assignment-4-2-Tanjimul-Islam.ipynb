```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv('/kaggle/input/datasets/lovishbansal123/titanic-dataset/Titanic-Dataset.csv')
iris = pd.read_csv('/kaggle/input/datasets/himanshunakrani/iris-dataset/iris.csv')
tips = pd.read_csv('/kaggle/input/datasets/saurabhbadole/restaurant-tips-dataset/tips.csv')
```

# Visualize at least 1 graph of each category. You can take help from the reference notebook

# Univariate Analysis

# 1. Pie chart

```python
class_count = titanic['Pclass'].value_counts().sort_index()
class_label = ['1st Class', '2nd Class', '3rd Class']

plt.figure(figsize = (7, 7))

plt.pie(class_count,
        labels = class_label,
        autopct = "%0.2f%%",
        colors = ['grey', 'orange', 'brown'],
        wedgeprops = {'width': 0.5})

plt.title('Passenger Class Distribution')
plt.show()
```

# 2. Count plot

```python
plt.figure(figsize = (7, 5))

sns.countplot(data = titanic, x = 'Sex')

plt.title('Number of Male and Female Passengers')
plt.xlabel('Gender')
plt.ylabel('Number of Passengers')
plt.grid()

plt.show()
```

# 3. Kde + Rug plot

```python
plt.figure(figsize = (8, 5))

sns.kdeplot(data = iris, x = 'petal_length')
sns.rugplot(data = iris, x = 'petal_length')

plt.title('Distribution of Iris Petal Length')
plt.xlabel('Petal Length')
plt.ylabel('Density')

plt.show()
```

# 4. Violin plot

```python
plt.figure(figsize = (7, 5))

sns.violinplot(data = titanic,
               x = 'Survived',
               y = 'Age')

plt.title('Age Distribution by Survival')
plt.xlabel('Survived')
plt.ylabel('Age')

plt.show()
```

# 5. Box plot

```python
plt.figure(figsize = (7, 5))

sns.boxplot(data = titanic,
            x = 'Pclass',
            y = 'Fare')

plt.title('Fare Distribution by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')

plt.show()
```

# 6. Boxen plot

```python
plt.figure(figsize = (8, 5))

sns.boxenplot(data = titanic,
              x = 'Pclass',
              y = 'Fare')

plt.title('Detailed Fare Distribution by Class')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')

plt.show()
```

# 7. Histogram

```python
data = titanic['Age'].dropna()

plt.figure(figsize = (8, 5))

plt.hist(data,
         bins = 15,
         color = 'brown',
         alpha = 0.75,
         edgecolor = 'black',
         cumulative = False)

plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Number of Passengers')

plt.show()
```

# Bivariate Analysis

# 1. Scatter plot

```python
plt.figure(figsize = (8, 5))

sns.scatterplot(data = tips,
                x = 'total_bill',
                y = 'tip',
                hue = 'day',
                size = 'size',
                style = 'smoker')

plt.title('Relationship Between Total Bill and Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')

plt.show()
```

# 2. Line plot

```python
plt.figure(figsize = (8, 5))

plt.plot(iris.index,
         iris['petal_length'])

plt.title('Petal Length Across Iris Samples')
plt.xlabel('Sample Number')
plt.ylabel('Petal Length')

plt.grid()

plt.show()
```

# 3. Joint plot

```python
sns.jointplot(data = tips,
              x = 'total_bill',
              y = 'tip')

plt.show()
```

# 4. Hexbin plot

```python
plt.figure(figsize = (8, 5))

plt.hexbin(tips['total_bill'],
           tips['tip'],
           gridsize = 15)

plt.colorbar()

plt.title('Total Bill and Tip Density')
plt.xlabel('Total Bill')
plt.ylabel('Tip')

plt.show()
```

# 5. 2D kde plot

```python
plt.figure(figsize = (8, 5))

sns.kdeplot(data = tips,
            x = 'total_bill',
            y = 'tip',
            fill = True)

plt.title('Density of Total Bill and Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')

plt.show()
```

# 6. Box plot

```python
plt.figure(figsize = (8, 5))

sns.boxplot(data = tips,
            x = 'day',
            y = 'tip')

plt.title('Tip Distribution by Day')
plt.xlabel('Day')
plt.ylabel('Tip')

plt.show()
```

# 7. Violin plot

```python
plt.figure(figsize = (8, 5))

sns.violinplot(data = tips,
               x = 'day',
               y = 'total_bill')

plt.title('Total Bill Distribution by Day')
plt.xlabel('Day')
plt.ylabel('Total Bill')

plt.show()
```

# 8. Swarm plot

```python
plt.figure(figsize = (8, 5))

sns.swarmplot(data = tips,
              x = 'day',
              y = 'tip')

plt.title('Individual Tips by Day')
plt.xlabel('Day')
plt.ylabel('Tip')

plt.show()
```

# 9. Count plot

```python
plt.figure(figsize = (8, 5))

sns.countplot(data = titanic,
              x = 'Pclass',
              hue = 'Survived')

plt.title('Survival Count by Passenger Class')
plt.xlabel('Passenger Class')
plt.ylabel('Number of Passengers')

plt.show()
```

# Multivariate Analysis

# 1. Scatter plot

```python
plt.figure(figsize = (9, 6))

sns.scatterplot(data = tips,
                x = 'total_bill',
                y = 'tip',
                hue = 'sex',
                size = 'size',
                style = 'smoker')

plt.title('Restaurant Bill Analysis')
plt.xlabel('Total Bill')
plt.ylabel('Tip')

plt.show()
```

# 2. Pair plot

```python
sns.pairplot(iris,hue = 'species')

plt.show()
```

# 3. Heatmap

```python
sns.heatmap(iris.select_dtypes(include = 'number').corr(),
            annot = True,
            cmap = 'coolwarm')

plt.title('Iris Feature Correlation')

plt.show()
```

# 4. FacetGrid plot

```python
g = sns.FacetGrid(tips,col = 'day',row = 'time')
g.map(sns.histplot,'tip')

plt.show()
```

# 5. PairGrid

```python
g = sns.PairGrid(iris,hue = 'species')

g.map_diag(sns.histplot)
g.map_upper(sns.scatterplot)
g.map_lower(sns.kdeplot)

plt.show()
```

# 6. Joint plot

```python
sns.jointplot(data = tips,x = 'total_bill',y = 'tip',hue = 'sex')

plt.show()
```

# 7. Boxen plot

```python
plt.figure(figsize = (9, 5))

sns.boxenplot(data = tips,
              x = 'day',
              y = 'total_bill',
              hue = 'sex')

plt.title('Total Bill Distribution by Day and Gender')
plt.xlabel('Day')
plt.ylabel('Total Bill')

plt.show()
```
