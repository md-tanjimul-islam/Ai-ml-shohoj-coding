# find minimun and maximum tip

```python
print(f"Min Tip: {np.min(tips['tip'].to_numpy())}, Max Tip: {np.max(tips['tip'].to_numpy())}")
```

# find average total_bill and average tip

```python
print(f"Average Total Bill: {np.mean(tips['total_bill'].to_numpy()):.2f}")
print(f"Average Tip: {np.mean(tips['tip'].to_numpy()):.2f}")
```

# find median total_bill

```python
print(f"Median Total Bill: {np.median(tips['total_bill'].to_numpy())}")
```

# find standard deviasion and variance of tip

```python
print(f"standard deviasion: {np.std(tips['tip'].to_numpy()):.2f}")
print(f"standard variance: {np.var(tips['tip'].to_numpy()):.2f}")
```

# apply round, floor and ceil operation on tip array

```python
print(f"rounded tip: {np.round(tips['tip'].to_numpy())}")
print(f"floored tip: {np.floor(tips['tip'].to_numpy())}")
print(f"ceiled tip: {np.ceil(tips['tip'].to_numpy())}")
```
