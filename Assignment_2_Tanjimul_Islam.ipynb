### Exercise 4.3: Early Stopping Mechanism (`break` and `continue`)
You are iterating through validation losses across 7 checkpoints:
`val_losses = [0.88, 0.82, 0.79, 0.65, 0.38, 0.32, 0.28]`

**Tasks:**
Loop through `val_losses`:
1. If loss is greater than `0.80`, skip reporting it using `continue`.
2. If loss drops below `0.40`, print `f"Early stopping triggered at loss {loss}! Model converged."` and terminate the loop using `break`.
3. Otherwise, print `f"Checkpoint validation loss: {loss}"`.

**Expected Output:**
```text
Checkpoint validation loss: 0.79
Checkpoint validation loss: 0.65
Early stopping triggered at loss 0.38! Model converged.
```


````python
val_losses = [0.88, 0.82, 0.79, 0.65, 0.38, 0.32, 0.28]

for loss in val_losses:
    if loss > 0.80:
        continue
    if loss < 0.40:
        print(f"Early stopping triggered at loss {loss}! Model converged.")
        break
    print(f"Checkpoint validation loss: {loss}")
````

### Exercise 9.1: Safe Feature Division (`ZeroDivisionError`)
In feature engineering, dividing two numeric features can crash if the denominator is zero.
Write a function `safe_feature_ratio(numerator, denominator)`:
- In `try`: returns `round(numerator / denominator, 2)`
- In `except ZeroDivisionError`: prints `"Warning: Denominator is 0. Returning 0.0"` and returns `0.0`
- Test with `safe_feature_ratio(150, 3)` and `safe_feature_ratio(150, 0)`.

**Expected Output:**
```text
Ratio 1: 50.0
Warning: Denominator is 0. Returning 0.0
Ratio 2: 0.0
```


````python
def safe_feature_ratio(numerator, denominator):
    try:
        return round(numerator / denominator, 2)
    except ZeroDivisionError:
        print("Warning: Denominator is 0. Returning 0.0")
        return 0.0

print(f"Ratio 1: {safe_feature_ratio(150, 3)}")
print(f"Ratio 2: {safe_feature_ratio(150, 0)}")
````

### Exercise 10.3: Server Access Log Extraction with Regex (`re`)

**Tasks:**
1. Import `re`.

3. Print the list of extracted items.

**Expected Output:**
